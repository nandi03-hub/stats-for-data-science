# The Dataset

Throughout this notebook we'll use the same classroom example from the accompanying Medium article.
Scores: 40, 60, 70, 80, 100

Average: 70

Using the same dataset allows us to connect the mathematics, the article, and the implementation.


```python
scores = [40,60,70,80,100]
scores
```

```python
total = 0

for score in scores:
    total = total + score

mean = total / len(scores)
print(mean)
```

## Step 1 — Compute the Deviations

Variance isn't interested in the values themselves.

It's interested in how far each observation lies from the mean.

Mathematically,

Deviation = Observation − Mean

```python
deviations = []

for score in scores: 
    deviations.append(score - mean)

deviations
```

Notice something interesting.

The deviations sum to zero.

That's why our first attempt at measuring spread failed.

Positive and negative deviations cancel each other out.

This is exactly the problem that motivated variance.

```python
total_deviation = 0

for deviation in deviations:
    total_deviation = total_deviation + deviation

print(total_deviation)
```

# Step 2 — Our First Attempt Failed

In the previous section, we discovered that simply adding the deviations gives zero.

That doesn't mean there is no variability.

It simply means positive and negative deviations cancel each other out.

So we need a different idea.

How can we measure distance from the mean without allowing positive and negative values to cancel?

There are two natural choices.

1. Ignore the signs by taking the absolute value.
2. Square every deviation.

Let's explore both approaches.

```python
absolute_deviations = []

for deviation in deviations:
    absolute_deviations.append(abs(deviation))

absolute_deviations
```

```python
total_absolute = 0

for value in absolute_deviations:
    total_absolute = total_absolute + value

mean_absolute_deviation = total_absolute/len(absolute_deviations)
print(f"Mean Absolute Deviation = {mean_absolute_deviation}")
```

## A Quick Detour: Mean Absolute Deviation (MAD)

Interestingly, this isn't a wrong approach.

In fact, we've just calculated another legitimate measure of spread called the **Mean Absolute Deviation (MAD).**

So why don't statisticians use it everywhere?

Because squaring the deviations gives variance some remarkable mathematical properties that make many statistical methods possible.

We'll come back to this idea later in the series.

For now, let's follow the path that eventually led to variance.

# Step 3 — Squaring the Deviations

Instead of removing the signs, let's square every deviation.

This immediately solves two problems.

- Every squared value is non-negative.
- Observations that are far from the mean contribute much more than observations that are close.

Let's see what happens.

```python
squared_deviations = []

for deviation in deviations:
    squared_deviations.append(deviation ** 2)

squared_deviations
```

```python
total_squared_deviation = 0

for value in squared_deviations:
    total_squared_deviation = total_squared_deviation + value

print(total_squared_deviation)
```

Now we have the total squared deviation.

But we're not interested in the total.

Imagine comparing two datasets.

One contains 5 observations.

Another contains 500 observations.

Naturally, the second dataset would have a much larger total simply because it has more data.

To make comparisons fair, we calculate the **average** squared deviation instead.

That's the final step.

```python
population_variance = total_squared_deviation / len(scores)
print(population_variance)
```

Congratulations!

We've just implemented **population variance** from first principles.

Notice something remarkable.

At no point did we memorise the variance formula.

We simply followed a sequence of logical steps:

1. Compute the mean.
2. Measure each observation's distance from the mean.
3. Square those distances.
4. Add them together.
5. Divide by the number of observations.

The mathematics naturally became an algorithm.

# Step 4 — Turning Mathematics into a Reusable Function

So far, we've calculated the variance for one specific dataset.

But what if we wanted to calculate the variance for **any** dataset?

Instead of rewriting the same code every time, we can package the algorithm into a reusable Python function.

This is one of the most important ideas in programming.

A function allows us to encapsulate logic once and reuse it many times.

## Designing the Function

Before writing any code, let's answer a few design questions.

### Input

A list of numerical values.

Example

[40, 60, 70, 80, 100]

### Output

A single number representing the population variance.

### Algorithm

1. Compute the mean.
2. Compute each deviation.
3. Square every deviation.
4. Sum the squared deviations.
5. Divide by the number of observations.

```python
def population_variance(data):
    """
    Calculate the population variance of a list of numbers.

    Parameters
    ----------
    data : list
        List of numerical observations.

    Returns
    -------
    float
        Population variance.
    """
    total = 0
    if len(data) == 0:
        raise ValueError("Dataset cannot be empty.")

    for value in data:
      total += value

    mean = total / len(data)

    squared_sum = 0
    for value in data:
       squared_sum += (value - mean) ** 2
    
    variance = squared_sum / len(data)

    return variance
```

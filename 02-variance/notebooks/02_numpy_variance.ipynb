# Variance Using NumPy

In the previous notebook, we built variance from scratch using pure Python.

That helped us understand the mathematics.

In practice, however, data scientists rarely calculate variance manually.

Instead, we rely on scientific computing libraries such as NumPy and Pandas.

In this notebook, we'll learn how to use those libraries correctly and more importantly, understand the subtle differences between them.

```python
import numpy as np 
import pandas as pd
```

```python
scores = np.array([40,60,70,80,100])
scores
```

## Population Variance

By default, NumPy computes the **population variance**.

Mathematically,

$$\sigma^2 = \frac{1}{N}\sum(x_i - \mu)^2$$

Let's verify it.

```python
np.var(scores)
```

## Sample Variance

Most real-world datasets are samples rather than complete populations.

NumPy allows us to compute sample variance using the `ddof` parameter.

```python
np.var(scores, ddof=1)
```

Notice something.

The sample variance is slightly larger.

That's because we're estimating the variability of an entire population using only a sample.

We'll derive why this correction works in a future notebook.

```python
series = pd.Series(scores)
series.var()
```

## Why Are They Different?

This surprises almost everyone the first time they encounter it.

NumPy assumes you're working with a **population**, so it divides by \(N\).

Pandas assumes you're working with a **sample**, so it divides by \(N-1\).

Neither library is wrong.

They're simply answering slightly different statistical questions.

# Why Do We Use NumPy?

By this point, you might be wondering:

> "If we've already implemented variance ourselves, why use NumPy at all?"

For small datasets, there isn't much difference.

But real-world datasets often contain hundreds of thousands or even millions of observations.

Scientific libraries like NumPy are implemented in highly optimized C code, making them dramatically faster than pure Python loops.

Let's see this in action.

```python
import random
import time
```

```python
# We Generate one million observations
large_dataset = [random.uniform(0,100) for _ in range (1_000_000)]

numpy_dataset = np.array(large_dataset)
```

```python
def population_variance(data):
    """
    Calculate the population variance of a list of numbers.

    Parameters
    ----------
    data : list
        List of numerical observations.

    Returns
    -------
    float
        Population variance.
    """
    total = 0
    if len(data) == 0:
        raise ValueError("Dataset cannot be empty.")

    for value in data:
      total += value

    mean = total / len(data)

    squared_sum = 0
    for value in data:
       squared_sum += (value - mean) ** 2
    
    variance = squared_sum / len(data)

    return variance
```

```python
#Manual implementation

start = time.perf_counter()

manual_result = population_variance(large_dataset)

manual_time = time.perf_counter() - start

#NumPy implementation

start = time.perf_counter()

numpy_result = np.var(numpy_dataset)

numpy_time = time.perf_counter() - start

print(f"Manual Variance : {manual_result:.6f}")
print(f"NumPy Variance : {numpy_result:.6f}")

print()

print(f"Manual Time : {manual_time:.4f} seconds")
print(f"NumPy Time : {numpy_time:.4f} seconds")

```

## What Did We Learn?

Both implementations produce essentially the same result.

That's reassuring—it means our mathematical implementation is correct.

The real difference lies in **speed**.

For educational purposes, implementing variance from scratch is invaluable because it helps us understand the mathematics.

In production environments, however, we almost always rely on optimized libraries such as NumPy because they are significantly faster and thoroughly tested.

Understanding the mathematics allows us to trust—and correctly use—the tools that data scientists rely on every day.

## Key Takeaways

In this notebook, we learned that:

- NumPy computes population variance by default.
- Pandas computes sample variance by default.
- The `ddof` parameter controls which formula NumPy uses.
- Scientific libraries should never be treated as black boxes.
- Understanding the underlying mathematics makes it much easier to use these tools correctly.
